```python
import tensorflow as tf # used for training ml models 
import matplotlib.pyplot as plt # for data visualisation 
import pandas as pd # for data manipulation and analysis 
import seaborn as sns # used for making stats graph
```

```python
# copied from keras image data loading 
training_set= tf.keras.utils.image_dataset_from_directory(
    'train', #directory name (location of data set)
    labels="inferred",  # labels are generated from directory structure (means directory se name copy krlo )
    label_mode="categorical", # means that the labels are encoded as a categorical vector(more than 2 classes of 2 use binary )
    class_names=None,
    color_mode="rgb",
    batch_size=32, # can be changed to 64 or 128 for speeding up model
    image_size=(128, 128), 
    shuffle=True, #  it will randomly select data.
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,

    
)
```

```python
validation_set= tf.keras.utils.image_dataset_from_directory(
    'valid', # directory name (location of data set)
    labels="inferred", # labels are generated from directory structure (means directory se name copy krlo )
    label_mode="categorical", # means that the labels are encoded as a categorical vector
    class_names=None,
    color_mode="rgb",
    batch_size=32, # can be changed to 64 or 128 for speeding up model(at a time 32 images and 32 labels  will be feeded to neural network)
    image_size=(128, 128),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
   
)
```

```python
# structure of training_set
for x,y in training_set:
    print(x,x.shape) # shape :  32 images 128*128 size and 3 as rgb color
    print(y,y.shape)
    break
```

```python
# Convolution neural network 
# it is divided into three process:
# 1.Convolution layer : input first goes to convolution layer where image is divided into pixels and
# image are treated as 2D matrix form and there is a filter which has some dimension and it scans(means note down important things
# from the image and save it to new matrix ) and moves to next area
# and scans the overall image and the output after all this is called feature map or activation map(many activation maps )
# 2.Pooling layer : after convolution image is passed to pooling layer .pooling layer is a downsampling operation where two filters 
# or parameter extracts values (max and avg ) : max selects the max value from a 2*2 cross filter and save it to new matrix 
# and avg similarily selects avg value from the matrix and save it to new matrix ...
# 3.Fully Connected layer : flattening and then flattened input where a single flattened input is connected to each neurons 
# flatenning converts multi dimensional feature map into one dimensional vector 

# Keras is a high-level API for building and training deep learning models.

# Overfitting : when your ml model learn so much that its performance is decreased when a new data is given 
# Overshooting : In machine learning, overshooting happens when a model's adjustments are too large. It's like taking too big of a step in the wrong direction. This can cause the model to miss the optimal solution and take longer to learn.
```

```python
from tensorflow.keras.layers import Dense,Conv2D,MaxPool2D,Flatten,Dropout
from tensorflow.keras.models import Sequential
```

## To avoid Overshooting :
### 1.Choose a small learning rate 0.0001 instead of 0.001(Default)
### 2. There may be chances of overfitting so increase number of neurons 
### 3. Add more convolution layer to extract more features from the images


```python
model =Sequential() # intialize sequential model (linear stack of layer )
```

```python
model.add(Conv2D(filters=32,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=32,kernel_size=3,activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
# kernel size = filter matrix size 
# input shape is the size of image and 3 is the rgb
# 32 different filters and each filter detect specific pattern or 32 different feature map
```

```python
# Conv2D: This layer performs 2D convolution.
# filters=32: This means the layer will learn 32 different filters
# kernel_size=3: The filter size is 3x3 pixels.
# padding='same': output and input image has same dimesion
# activation='relu': The ReLU (Rectified Linear Unit) activation function is applied to introduce non-linearity into the network

# MaxPooling: This layer downsamples the feature maps, reducing their spatial dimensions.
# pool_size=2: The pooling window is 2x2 pixels.
# strides=2: The window slides over the input with a stride of 2 pixels
```

```python
model.add(Conv2D(filters=64,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=64,kernel_size=3,activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=128,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=128,kernel_size=3,activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=256,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=256,kernel_size=3,activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
model.add(Conv2D(filters=512,kernel_size=3,padding='same',activation='relu',input_shape=[128,128,3]))
model.add(Conv2D(filters=512,kernel_size=3,activation='relu'))
model.add(MaxPool2D(pool_size=2,strides=2))
```

```python
#model.add(Dropout(0.25)) ##to void overfitting
```

```python
model.add(Flatten())
```

```python
model.add(Dense(units=1500,activation='relu'))
```

```python
#model.add(Dropout(0.4)) # ignores a fraction of neurons in a layer 
```

```python
# output layer 38 different output or neureons 
model.add(Dense(units=38,activation='softmax')) #softmax:-Outputs probabilities for each disease class.

```

```python
import os

os.environ["TF_USE_LEGACY_KERAS"] = "True"

model.compile(optimizer=tf.keras.optimizers.legacy.Adam(
    learning_rate=0.0001),loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
# optimizers are algorithms that adjust the weights and biases of a neural network to minimize the loss function.
```

```python
model.summary()
```

```python
training_history=model.fit(x=training_set,validation_data=validation_set,epochs=10)
```

## Model Evaluation

```python
# Model Evaluation on training set 
train_loss,train_acc = model.evaluate(training_set)
```

```python
print(train_loss,train_acc)
```

```python
# Model Evaluation on validation set
val_loss,val_acc = model.evaluate(validation_set)
```

```python
print(val_loss,val_acc)
```

```python
y_pred = model.predict(test_set)
y_pred,y_pred.shape # for each images it is providing 38 probabilities
```

```python
predicted_categories = tf.argmax(y_pred , axis = 1)# axis = 1 means search by column wise 
# this function return the max probability index 
```

```python
predicted_categories
```

```python
true_categories = tf.concat([y for x , y in test_set], axis = 0) # concat all the y in xy test in row wise manner 
true_categories
```

```python
# 1,0,0,0.....38 class means this belong to class 1 
```

```python
Y_true = tf.argmax(true_categories , axis = 1 )
Y_true
```

```python
# Precision and Recall in machine learning 
# Precision measures the percentage of predictions made by the model that are correct
# (how accurate your search is)
# Recall measures the percentage of relevant data points that were correctly identified by the model

# Recall :- true positive out of total positive in real (kitne plants ko model ne shi predict kiya out of total shi real  plants )
# Precision :- true postive out of total positive in prediction 
```

```python
from sklearn.metrics import classification_report,confusion_matrix # from the documentation of sklearn.metrices
```

```python
print(classification_report(Y_true,predicted_categories,target_names = class_name))
```

```python
# F1 score is harmonic mean of precision and recall values
```

```python
cm = confusion_matrix(Y_true,predicted_categories) # from the documentation of sklearn.metrices 
cm,cm.shape # confusion matrix of 38*38
```

```python
plt.figure(figsize = (40,40))
sns.heatmap(cm,annot = True,annot_kws = {'size':10}) # from the documentation of seaborn
plt.xlabel("Predicted Class",fontsize = 20)
plt.ylabel("Actual Class" , fontsize = 20)
plt.title("Plant Disease Prediction Confusion Matrix",fontsize = 25)
plt.show()
```
