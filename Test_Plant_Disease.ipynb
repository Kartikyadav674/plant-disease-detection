### Importing Libraries

```python
import numpy as np
import tensorflow as tf

import matplotlib.pyplot as plt
```

### Loading Model

```python
model = tf.keras.models.load_model('trained_model.h5')
```

```python
model.summary()
```

```python
import cv2 # for visualization 
image_path = "test/test/CornCommonRust2.JPG"
# Reading Image 
img = cv2.imread(image_path)
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB) # CONVERT BGR IMG TO RGB



# Displaying Image
plt.imshow(img)
plt.title("Test Image")
plt.xticks([])
plt.yticks([])
plt.show()
```

### Testing Model

```python
image = tf.keras.preprocessing.image.load_img(image_path,target_size = (128,128)) # our model take the 128*128 size of image 
input_arr = tf.keras.preprocessing.image.img_to_array(image)                      # in array format 
input_arr = np.array([input_arr]) # convert single image to a batch
print(input_arr.shape)
```

```python
prediction = model.predict(input_arr)
prediction,prediction.shape # 1 row and 38 column matrix
# each column represent for each classes
# what is the probability that this image belongs to class 1
```

```python
# we will retrun the class with max probability 
result_index = np.argmax(prediction)
result_index
```
